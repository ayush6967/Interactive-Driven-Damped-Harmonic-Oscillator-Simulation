#  Driven Damped Harmonic Oscillator Simulation


### What is a Harmonic Oscillator?
A harmonic oscillator is any physical system where a restoring force pulls it back 
to equilibrium proportional to its displacement. The simplest example: a mass on a spring. 
Pull it and it snaps back, overshoots, and oscillates. This same mathematics describes 
pendulums, electric circuits, molecular vibrations, and gravitational waves.

The equation of motion is:
**mẍ = -kx - bẋ + F₀cos(ω_d t)**

Where:
- m = mass, k = spring constant, b = damping coefficient
- F₀ = driving force amplitude, ω_d = driving frequency
- x = displacement, ẋ = velocity, ẍ = acceleration

### What is Damping?
In reality, friction and air resistance drain energy from the system. 
A damped oscillator loses amplitude over time and eventually stops, unless it is driven externally.
The damping coefficient b controls how fast energy is lost.

### What is a Driving Force?
An external periodic force applied to the system, like pushing a swing at regular intervals.
This continuously pumps energy in, competing against damping.

### What is Resonance?
Every oscillator has a natural frequency ω₀ = √(k/m). The frequency at which it naturally wants 
to oscillate at. When the driving frequency matches this, energy builds up continuously 
and the amplitude grows dramatically. This is resonance.

Real-world resonance:
- Tacoma Narrows Bridge (1940): wind drove the bridge at its natural frequency until it collapsed.
- MRI machines: hydrogen atoms are driven at their natural frequency to image the body.
- Musical instruments: strings and air columns resonate at specific frequencies.
- Microwave ovens: water molecules are driven at their natural frequency to heat food.

## Suggested Experiments For You to Play Around with this Program !!

Here are some ideas to explore the physics. Change parameters and observe how the system responds:

### 1. Pure Simple Harmonic Motion (no damping, no driving)
- Set b = 0, F0 = 0
- This is the ideal and simple textbook case, the mass oscillates forever.
- Phase space will show a perfect ellipse.
- Total energy stays perfectly constant.

### 2. Damping Only (no driving force)
- Set F0 = 0, keep b > 0 (any positive value).
- Watch amplitude decay over time.
- Phase space shows an inward spiral — energy draining away.
- Try different values of b: higher b means faster decay and vice versa.

### 3. Witness Resonance
- Set b = 0.05 (or any other low damping value), F0 = 2, t_max = 300.
- Type 'resonance' when asked for driving frequency.
- Watch amplitude build continuously on the Position vs Time plot.
- Watch energy climb to hundreds of Joules on the Energy plot.
- Phase space shows a beautiful outward spiral.
- Here, in the damping (b) vs external driving force (F0) competition, the external driving force is winning.

### 4. Compare Resonance vs Off-Resonance
- Run once at resonance, note max displacement and max energy.
- Run again with driving frequency (ω_d) very different from natural frequency (ω₀). For example: try ω_d = 1 when ω₀ ≈ 3.16.
- The difference in amplitude and energy is dramatic, often 30-100x even.

### 5. Effect of Damping on Resonance
- Run at resonance with b = 0.5 (heavy damping)
- Run at resonance with b = 0.05 (light damping)
- Light damping allows much higher resonance amplitude
- Heavy damping suppresses resonance significantly

### 6. Simulate the Tacoma Narrows Bridge
- Set m = 1000, k = 100, b = 0.1, F0 = 50, t_max = 300
- Type 'resonance' when asked for driving resonance frequency.
- Watch energy build until the system is wildly oscillating — that's what happened to the bridge.

### 7. Beats Phenomenon
- Set b = 0.01 (very low damping), F0 = 2, t_max = 300
- Enter ω_d slightly different from ω₀ (you can try ω₀ + 0.5).
- Energy plot shows rhythmic rising and falling, that's beats.
- Two close frequencies are interfering with each other.


## What the Plots Show

**Plot 1: Position vs Time (x vs t)**
How the mass moves over time. At resonance you'll see amplitude growing continuously.
Off resonance you'll see small constant oscillations.

**Plot 2: Position vs Velocity (x vs v)**
Every point represents the complete state of the system at one instant.
- Perfect circle plot = undamped SHM
- Inward spiral plot = damping killing the motion
- Outward spiral plot = resonance building energy
- Tangled complex plot = off-resonance, two frequencies fighting

**Plot 3: Energy vs Time (E vs t)**
Kinetic energy (blue), potential energy (orange) and total energy (green dashed).
Watch kinetic and potential constantly converting into each other while total energy 
either grows (resonance), stays constant (no damping), or decays (damping only).

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp
```

```python
# Variable naming convention:
# m = mass, k = spring constant, b = damping coefficient
# F0 = driving amplitude, omega_0 = natural frequency, omega_d = driving frequency
# t_max = total simulation duration (seconds), t = time array from 0 to t_max

m = float(input("Enter mass (kg): "))
k = float(input("Enter spring constant (N/m) [higher value means stiffer spring]: "))
b = float(input("Enter damping coefficient (kg/s) [higher value means more friction/energy loss]: "))
F0 = float(input("Enter driving force amplitude (N) [this is the strength of the external push]: "))
t_max = float(input("Enter simulation duration (seconds) [how long do you want to observe the system]: "))

# Calculate and display natural frequency
omega_0 = np.sqrt(k / m)
print(f"\n--- System Information so far ---")
print(f"Natural frequency of your system: ω₀ = {omega_0:.4f} rad/s")
print(f"This is the frequency at which your system naturally WANTS to oscillate at.")
print(f"\nYou can now choose the frequency at which to drive (push) the system externally.")
print(f"If the driving frequency matches the natural frequency, the system enters RESONANCE")
print(f"A state where energy builds up dramatically and amplitude grows very large.")
print(f"\nType 'resonance' to drive at ω_d = ω₀ = {omega_0:.4f} rad/s and witness the system at resonance.")
print(f"Or enter any other value of your choice to see how the system behaves off-resonance.\n")

# User chooses driving frequency
user_input = input("Enter driving frequency (rad/s) or type 'resonance': ")

if user_input.lower() == "resonance" :
    omega_d = omega_0
    print(f"\nDriving the system at resonance: ω_d = ω₀ = {omega_0:.4f} rad/s")
    print("Now witness the system at RESONANCE and see how the amplitude grows over time!")
else:
    omega_d = float(user_input)
    print(f"\nDriving the system at ω_d = {omega_d:.4f} rad/s")
    print(f"While the Natural frequency was, ω₀ = {omega_0:.4f} rad/s")
    if omega_d < omega_0:
        print("Driving below natural frequency, system will respond but not dramatically.")
    elif omega_d > omega_0:
        print("Driving above natural frequency, system will struggle to keep up.")

# Time array
t = np.linspace(0, t_max, 3000)
```

```python
def harmonic_oscillator(t, state):
    x, v = state
    dxdt = v
    dvdt = (-k * x - b * v + F0 * np.cos(omega_d * t)) / m
    return [dxdt, dvdt]
```

```python
# Initial conditions — system starts at rest at equilibrium
x0 = 1.0   # initial position (m)
v0 = 0.0   # initial velocity (m/s)
initial_state = [x0, v0]

# Run the solver
solution = solve_ivp(
    harmonic_oscillator,
    [0, t_max],
    initial_state,
    t_eval=t,
    max_step=0.01
)

# Extract results
position = solution.y[0]
velocity = solution.y[1]
```

```python
# x vs t pLot
plt.figure(figsize=(12, 5))

plt.plot(t, position, color="steelblue", linewidth=1.5)

plt.title(f"Driven Damped Harmonic Oscillator\n"
          f"ω₀ = {omega_0:.4f} rad/s | ω_d = {omega_d:.4f} rad/s | "
          f"b = {b} | F0 = {F0}")
plt.xlabel("Time (seconds)")
plt.ylabel("Position (m)")
plt.axhline(0, color="gray", linewidth=0.8, linestyle="--")
plt.grid(True)
plt.tight_layout()
plt.show()

print(f"\nMax displacement reached: {max(abs(position)):.4f} m")
```

```python
# x vs v Plot
plt.figure(figsize=(8, 6))

plt.plot(position, velocity, color="darkorange", linewidth=1)

plt.title(f"Phase Space (Position vs Velocity)\n"
          f"ω₀ = {omega_0:.4f} rad/s | ω_d = {omega_d:.4f} rad/s | "
          f"b = {b} | F0 = {F0}")
plt.xlabel("Position (m)")
plt.ylabel("Velocity (m/s)")
plt.axhline(0, color="gray", linewidth=0.8, linestyle="--")
plt.axvline(0, color="gray", linewidth=0.8, linestyle="--")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Energy Plot
kinetic_energy = 0.5 * m * velocity**2
potential_energy = 0.5 * k * position**2
total_energy = kinetic_energy + potential_energy

plt.figure(figsize=(12, 5))

plt.plot(t, kinetic_energy, color="steelblue", linewidth=1, label="Kinetic Energy")
plt.plot(t, potential_energy, color="darkorange", linewidth=1, label="Potential Energy")
plt.plot(t, total_energy, color="green", linewidth=1.5, label="Total Energy", linestyle="--")

plt.title(f"Energy vs Time\n"
          f"ω₀ = {omega_0:.4f} rad/s | ω_d = {omega_d:.4f} rad/s | "
          f"b = {b} | F0 = {F0}")
plt.xlabel("Time (seconds)")
plt.ylabel("Energy (J)")
plt.legend()
plt.grid(True)
plt.legend(loc="upper right")
plt.tight_layout()
plt.show()

print(f"Max total energy reached: {max(total_energy):.4f} J")
```
